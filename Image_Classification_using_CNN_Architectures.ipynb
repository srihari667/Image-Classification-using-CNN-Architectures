1.  What is a Convolutional Neural Network (CNN), and how does it differ from traditional fully connected neural networks in terms of architecture and performance on image data?


    **A Convolutional Neural Network (CNN)** is a specialized type of deep learning model designed to process grid-structured data such as images. CNNs automatically learn spatial features like edges, textures, and shapes using *convolutional filters*.

    ### **How CNNs work**

    CNNs use three major components:

    1. **Convolutional Layers**

       * Apply filters (kernels) that slide over the image.
       * Capture local patterns such as edges and corners.
       * Use *shared weights*, which greatly reduces parameter count.

    2. **Pooling Layers**

       * Downsample feature maps.
       * Reduce computational cost and add spatial invariance.

    3. **Fully Connected Layers**

       * Perform final classification based on learned features.

    # **Difference between CNNs and Fully Connected Neural Networks**

    | Feature                          | CNN                                          | Fully Connected Network                             |
    | -------------------------------- | -------------------------------------------- | --------------------------------------------------- |
    | **Connection type**              | Local (neighborhood-based)                   | Every neuron connects to all previous neurons       |
    | **Parameters**                   | Very few (due to weight sharing)             | Extremely large (high memory + compute)             |
    | **Preserves spatial structure?** | Yes                                          | No (image flattened)                                |
    | **Best for**                     | Images, videos, audio                        | Tabular data, simple classification                 |
    | **Performance on images**        | High accuracy, learns features automatically | Poor; easily overfits and computationally expensive |


    # **Why CNNs perform better on image data**

    * They **learn hierarchical features** automatically (edges → textures → objects).
    * **Local receptive fields** capture spatial relationships.
    * **Parameter sharing** reduces training time and avoids overfitting.
    * Better generalization on real-world images.



2.  Discuss the architecture of LeNet-5 and explain how it laid the foundation for modern deep learning models in computer vision. Include references to its original research paper.


    **LeNet-5**, proposed by *Yann LeCun et al. (1998)*, is one of the earliest and most influential Convolutional Neural Network (CNN) architectures. It was designed for handwritten digit recognition (MNIST dataset) and laid the foundation for modern deep learning in computer vision.

    ## **Architecture of LeNet-5**

    LeNet-5 takes a **32×32 grayscale image** as input and consists of **7 layers** (excluding input), including convolution, pooling, and fully connected layers.

    ### **Layer-by-Layer Structure**

    1. **Input Layer:**

       * 32×32×1 grayscale image.

    2. **C1 — Convolution Layer:**

       * 6 filters of size 5×5
       * Output: 28×28×6
       * Activation: **tanh**

    3. **S2 — Subsampling (Average Pooling):**

       * 2×2 pooling with stride 2
       * Output: 14×14×6

    4. **C3 — Convolution Layer:**

       * 16 filters of size 5×5
       * Output: 10×10×16
       * Uses a special partial connection pattern (not fully connected).

    5. **S4 — Subsampling Layer:**

       * 2×2 average pooling
       * Output: 5×5×16

    6. **C5 — Convolution Layer:**

       * 120 filters of size 5×5
       * Output: 1×1×120
       * (Equivalent to a fully connected layer)

    7. **F6 — Fully Connected Layer:**

       * 84 neurons
       * Activation: tanh

    8. **Output Layer:**

       * 10 neurons for 10 digit classes
       * Uses **Gaussian RBF** in the original architecture (modern versions use softmax)

    ## **How LeNet-5 Influenced Modern CNNs**

    LeNet-5 introduced fundamental ideas that shaped today’s deep learning models:

    ### **1. Local Receptive Fields**

    Instead of connecting every neuron to the entire input, each neuron sees only a small region (5×5), reducing complexity.

    ### **2. Weight Sharing**

    Convolutional filters share weights across the spatial dimension, drastically reducing parameters and making learning easier.

    ### **3. Hierarchical Feature Learning**

    Lower layers learn edges and blobs; deeper layers learn shapes — a concept used in AlexNet, VGG, ResNet, etc.

    ### **4. Subsampling (Pooling)**

    Pooling layers reduce spatial size, control overfitting, and make the network robust to small shifts.

    ### **5. End-to-End Learning**

    LeNet-5 demonstrated that neural networks could learn features **directly from raw pixels**, influencing modern image recognition systems.

3.  Compare and contrast AlexNet and VGGNet in terms of design principles,
number of parameters, and performance. Highlight key innovations and limitations of each.


    **AlexNet (2012)** and **VGGNet (2014)** are two landmark CNN architectures that significantly advanced deep learning for image classification. Both achieved high performance on the ImageNet dataset, but their design philosophies are very different.


    # **1. Design Principles**

    ## **AlexNet (2012)**

    * Introduced by Alex Krizhevsky et al.
    * **8 layers** deep (5 convolutional + 3 fully connected)
    * Used **large kernels** like 11×11 and 5×5 in early layers
    * Introduced several key innovations:

      * **ReLU activation** (instead of tanh)
      * **Dropout** to reduce overfitting
      * **Data augmentation**
      * **Training on GPU** (2 GPUs used)
      * **Local Response Normalization (LRN)**

    ## **VGGNet (2014) — VGG16/VGG19**

    * Proposed by Simonyan & Zisserman
    * **Very deep architecture** (16 or 19 layers)
    * Uses **only 3×3 convolution filters**
    * Follows a simple and uniform design:

      * Stack multiple 3×3 conv layers
      * Reduce spatial size using max pooling
    * No LRN, no fancy tricks — just depth + small filters


    # **2. Number of Parameters**

    | Model       | Approx. Parameters | Comments                       |
    | ----------- | ------------------ | ------------------------------ |
    | **AlexNet** | ~60 million        | Large due to big FC layers     |
    | **VGG16**   | ~138 million       | Much larger; heavy computation |
    | **VGG19**   | ~143 million       | Even more parameters           |

    VGGNet has **over 2× more parameters** than AlexNet, mainly because of deeper layers and large fully connected layers.


    # **3. Performance Comparison**

    * **AlexNet:**

      * Top-5 error ≈ 15% on ImageNet (2012)
      * First major GPU-trained CNN to beat traditional methods

    * **VGGNet:**

      * Top-5 error ≈ 7.3% (VGG16) — significantly better
      * Depth improved accuracy but increased computational cost


    # **4. Key Innovations**

    ### **AlexNet Innovations**

    * ReLU activation → solved vanishing gradient problems
    * Dropout → reduced overfitting
    * GPU training → enabled deeper models
    * Data augmentation → improved generalization

    ### **VGGNet Innovations**

    * Showed **depth** is the key to high accuracy
    * Small 3×3 filters → efficient feature extraction
    * Clean, uniform architecture → easy to extend
      (influenced ResNet, EfficientNet, etc.)

    # **5. Limitations**

    ### **AlexNet Limitations**

    * Large kernel sizes → heavy computation
    * LRN now considered unnecessary
    * Architecture not modular

    ### **VGGNet Limitations**

    * Very large model size (~138M parameters)
    * Slow to train, slow to deploy
    * High memory consumption (hard on small GPUs)

4. What is transfer learning in the context of image classification? Explain
how it helps in reducing computational costs and improving model performance with limited data.


    **Transfer learning** is a deep learning technique where a model trained on a large dataset (such as ImageNet with 1.2 million images) is reused as the starting point for a new, smaller task.

    In image classification, transfer learning means taking a **pre-trained CNN** (e.g., VGG16, ResNet50, InceptionV3) and adapting it to a new dataset that usually has fewer images.


    ## **How Transfer Learning Works**

    There are two common approaches:

    ### **1. Feature Extraction (Freeze Base Model)**

    * Use the convolutional layers of a pre-trained model as a **fixed feature extractor**.
    * Only train the newly added fully connected layers.
    * Useful when the new dataset is very small.

    ### **2. Fine-Tuning (Unfreeze Some Layers)**

    * Unfreeze the top few layers of the pre-trained model.
    * Train them along with the new classification head using a low learning rate.
    * Helps the model adapt better to target-domain features.


    # **Why Transfer Learning Helps**

    ### **Reduces Computational Cost**

    * Training large CNNs from scratch requires massive GPU time and huge datasets.
    * With transfer learning, most weights are already trained; only a small part is trained again.
    * Saves time and computation.

    ### **Better Performance With Limited Data**

    * Pre-trained models already learned low-level patterns:

      * Edges
      * Textures
      * Shapes
    * These features are universal and transfer well to new tasks.
    * Prevents **overfitting** when data is small.

    ### **Faster Convergence**

    * Model starts from a strong baseline instead of random initialization.
    * Needs fewer epochs to reach good accuracy.

5.  Describe the role of residual connections in ResNet architecture. How do they address the vanishing gradient problem in deep CNNs?


    **Residual connections**, introduced in the **ResNet (Residual Network)** architecture by He et al. (2015), are shortcut (skip) connections that allow the input of a layer to be added directly to the output of deeper layers.

    A residual block computes:

    ### **y = F(x) + x**

    Where:

    * **x** = input
    * **F(x)** = output of stacked convolution layers
    * **x** is added directly to bypass these layers.

    # **Why Residual Connections Are Important**

    ## **1. Solve the Vanishing Gradient Problem**

    In very deep networks, gradients become extremely small during backpropagation, causing the model to stop learning.
    Residual connections create a **direct gradient path**, allowing gradients to flow backward without shrinking.

    This enables stable training of **very deep models** (e.g., 50, 101, 152 layers).

    ## **2. Prevent Degradation in Deep Networks**

    Before ResNet, adding more layers sometimes *reduced* accuracy because deeper models were harder to optimize.

    Residual connections allow the model to learn:

    * **Residual mapping (F(x))** instead of the full mapping
    * If deeper layers don't help, they simply learn **F(x) = 0**, so the block becomes **identity mapping**

    This prevents performance degradation.

    ## **3. Faster and Easier Optimization**

    * Skip connections act like “information highways”
    * They stabilize training
    * Make optimization smoother
    * Allow deeper architectures without risk


```python
#6. Implement the LeNet-5 architectures using Tensorflow or PyTorch to classify the MNIST dataset. Report the accuracy and training time.

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
import numpy as np
from tensorflow.keras.utils import to_categorical

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

x_train = np.pad(x_train, ((0,0),(2,2),(2,2)), mode='constant')
x_test  = np.pad(x_test,  ((0,0),(2,2),(2,2)), mode='constant')

x_train = x_train.astype("float32")/255.0
x_test  = x_test.astype("float32")/255.0

x_train = x_train[...,None]
x_test  = x_test[...,None]

y_train = to_categorical(y_train, 10)
y_test  = to_categorical(y_test, 10)

def build_lenet5():
    model = keras.Sequential([
        layers.Conv2D(6, (5,5), activation='tanh', input_shape=(32,32,1)),
        layers.AveragePooling2D(pool_size=(2,2)),
        layers.Conv2D(16, (5,5), activation='tanh'),
        layers.AveragePooling2D(pool_size=(2,2)),
        layers.Conv2D(120, (5,5), activation='tanh'),
        layers.Flatten(),
        layers.Dense(84, activation='tanh'),
        layers.Dense(10, activation='softmax')
    ])
    return model

model = build_lenet5()
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

model.summary()

history = model.fit(x_train, y_train, epochs=10, batch_size=128, validation_split=0.1)


test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test Accuracy:", test_acc)


```

```python
'''
7. Use a pre-trained VGG16 model (via transfer learning) on a small custom
dataset (e.g., flowers or animals). Replace the top layers and fine-tune the model. Include your code and result discussion.
'''

import tensorflow as tf
from tensorflow.keras import layers, models, applications, optimizers, callbacks
import tensorflow_datasets as tfds
import matplotlib.pyplot as plt
import numpy as np
import os

IMG_SIZE = 224
BATCH = 32
HEAD_EPOCHS = 6
FT_EPOCHS = 5
SEED = 42

(ds_train, ds_val), info = tfds.load('tf_flowers',
                                     split=['train[:80%]', 'train[80%:]'],
                                     with_info=True, as_supervised=True)
NUM_CLASSES = info.features['label'].num_classes
print("Num classes:", NUM_CLASSES)

AUTOTUNE = tf.data.AUTOTUNE

def preprocess_train(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.image.random_flip_left_right(image)
    image = tf.image.random_brightness(image, 0.1)
    image = tf.image.random_contrast(image, 0.9, 1.1)
    image = image / 255.0
    return image, label

def preprocess_val(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = image / 255.0
    return image, label

train_ds = ds_train.map(preprocess_train, num_parallel_calls=AUTOTUNE).shuffle(1024, seed=SEED).batch(BATCH).prefetch(AUTOTUNE)
val_ds   = ds_val.map(preprocess_val, num_parallel_calls=AUTOTUNE).batch(BATCH).prefetch(AUTOTUNE)

base_model = applications.VGG16(weights='imagenet', include_top=False, input_shape=(IMG_SIZE, IMG_SIZE, 3))
base_model.trainable = False

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = applications.vgg16.preprocess_input(inputs * 255.0)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.4)(x)
outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)
model = models.Model(inputs, outputs)

model.compile(optimizer=optimizers.Adam(learning_rate=1e-4),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
model.summary()

ckpt_path = "vgg16_flowers_best.h5"
cb = [
    callbacks.ModelCheckpoint(ckpt_path, monitor='val_accuracy', save_best_only=True, verbose=1),
    callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True)
]

history_head = model.fit(train_ds, validation_data=val_ds, epochs=HEAD_EPOCHS, callbacks=cb)

plt.plot(history_head.history['accuracy'], label='train_acc_head')
plt.plot(history_head.history['val_accuracy'], label='val_acc_head')
plt.legend(); plt.show()

base_model.trainable = True
for layer in base_model.layers:
    if layer.name.startswith('block5'):
        layer.trainable = True
    else:
        layer.trainable = False

model.compile(optimizer=optimizers.Adam(learning_rate=1e-5),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history_ft = model.fit(train_ds, validation_data=val_ds, epochs=FT_EPOCHS, callbacks=cb)

plt.plot(history_head.history['val_accuracy'] + history_ft.history['val_accuracy'], label='val_acc_all')
plt.legend(); plt.show()

val_loss, val_acc = model.evaluate(val_ds)
print(f"Final validation accuracy: {val_acc:.4f}")

model.save("vgg16_flowers_final.h5")
print("Model saved to vgg16_flowers_final.h5")

class_names = info.features['label'].names
for images, labels in val_ds.take(1):
    preds = model.predict(images)
    preds_labels = np.argmax(preds, axis=1)
    for i in range(6):
        plt.subplot(2,3,i+1)
        plt.imshow(images[i].numpy())
        plt.title(f"GT: {class_names[int(labels[i])]} | Pred: {class_names[int(preds_labels[i])]}")
        plt.axis('off')
plt.show()

```

```python
'''
8. Write a program to visualize the filters and feature maps of the first
convolutional layer of AlexNet on an example input image.
'''

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, preprocessing

def build_alexnet_first(input_shape=(227,227,3)):
    inp = layers.Input(shape=input_shape)
    x = layers.Conv2D(96, (11,11), strides=4, activation='relu', name='conv1')(inp)
    return models.Model(inputs=inp, outputs=x)

alex1 = build_alexnet_first()
alex1.summary()

def show_filters(layer_weights, n_filters=12):
    w = layer_weights
    w_norm = (w - w.min()) / (w.max() - w.min() + 1e-8)
    plt.figure(figsize=(n_filters*1.2, 2))
    for i in range(n_filters):
        plt.subplot(1, n_filters, i+1)
        plt.imshow(w_norm[:,:,:,i])
        plt.axis("off")
    plt.suptitle("AlexNet Conv1 Filters (Random Init)")
    plt.show()

def show_feature_maps(feature_maps, n_maps=12):
    fmap = feature_maps[0]
    plt.figure(figsize=(n_maps*1.2, 2))
    for i in range(n_maps):
        plt.subplot(1, n_maps, i+1)
        plt.imshow(fmap[:,:,i], cmap='viridis')
        plt.axis("off")
    plt.suptitle("Conv1 Feature Maps")
    plt.show()

img_url = "https://tensorflow.org/images/blogs/serving/cat.jpg"
img_path = tf.keras.utils.get_file("cat.jpg", origin=img_url)

img = preprocessing.image.load_img(img_path, target_size=(227,227))
x = preprocessing.image.img_to_array(img)
x = np.expand_dims(x, 0) / 255.0

filters = alex1.get_layer("conv1").get_weights()[0]
show_filters(filters, n_filters=12)

feature_maps = alex1.predict(x)
show_feature_maps(feature_maps, n_maps=12)
```

```python
'''
9.  Train a GoogLeNet (Inception v1) or its variant using a standard dataset
like CIFAR-10. Plot the training and validation accuracy over epochs and analyze
overfitting or underfitting.
'''


import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
import numpy as np
import matplotlib.pyplot as plt
import os

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
y_train = y_train.ravel()
y_test = y_test.ravel()

x_train = x_train.astype("float32") / 255.0
x_test  = x_test.astype("float32")  / 255.0

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomTranslation(0.06, 0.06),
    layers.RandomZoom(0.06, 0.06),
], name="data_augmentation")

def inception_module(x, filters_1x1, filters_3x3_reduce, filters_3x3,
                     filters_5x5_reduce, filters_5x5, filters_pool_proj, name=None):
    branch1 = layers.Conv2D(filters_1x1, (1,1), padding="same", activation="relu")(x)
    branch2 = layers.Conv2D(filters_3x3_reduce, (1,1), padding="same", activation="relu")(x)
    branch2 = layers.Conv2D(filters_3x3, (3,3), padding="same", activation="relu")(branch2)
    branch3 = layers.Conv2D(filters_5x5_reduce, (1,1), padding="same", activation="relu")(x)
    branch3 = layers.Conv2D(filters_5x5, (3,3), padding="same", activation="relu")(branch3)
    branch3 = layers.Conv2D(filters_5x5, (3,3), padding="same", activation="relu")(branch3)
    branch4 = layers.MaxPool2D((3,3), strides=(1,1), padding="same")(x)
    branch4 = layers.Conv2D(filters_pool_proj, (1,1), padding="same", activation="relu")(branch4)
    out = layers.Concatenate(axis=-1, name=name)([branch1, branch2, branch3, branch4])
    return out

input_shape = x_train.shape[1:]
inputs = layers.Input(shape=input_shape)

x = data_augmentation(inputs)
x = layers.Conv2D(64, (3,3), padding="same", activation="relu")(x)
x = layers.BatchNormalization()(x)
x = layers.Conv2D(64, (3,3), padding="same", activation="relu")(x)
x = layers.MaxPool2D((2,2))(x)

x = inception_module(x, 32, 32, 32, 16, 16, 16, name="incept_1")
x = inception_module(x, 64, 48, 64, 16, 32, 32, name="incept_2")
x = layers.MaxPool2D((2,2))(x)

x = inception_module(x, 80, 56, 80, 24, 48, 32, name="incept_3")
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.4)(x)
outputs = layers.Dense(10, activation="softmax")(x)

model = models.Model(inputs, outputs, name="mini_googlenet_cifar10")
model.summary()

model.compile(
    optimizer=tf.keras.optimizers.Adam(),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

checkpoint_path = "model_checkpoint.h5"
callbacks = [
    tf.keras.callbacks.ModelCheckpoint(checkpoint_path, monitor="val_accuracy", save_best_only=True, verbose=1),
    tf.keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=8, restore_best_weights=True, verbose=1),
    tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=4, verbose=1)
]

EPOCHS = 50
BATCH_SIZE = 128

history = model.fit(
    x_train, y_train,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    validation_split=0.1,
    callbacks=callbacks,
    shuffle=True
)

acc = history.history.get("accuracy", [])
val_acc = history.history.get("val_accuracy", [])
loss = history.history.get("loss", [])
val_loss = history.history.get("val_loss", [])

epochs_range = range(1, len(acc) + 1)

plt.figure(figsize=(9,4))
plt.plot(epochs_range, acc, label="Training accuracy")
plt.plot(epochs_range, val_acc, label="Validation accuracy")
plt.title("Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(9,4))
plt.plot(epochs_range, loss, label="Training loss")
plt.plot(epochs_range, val_loss, label="Validation loss")
plt.title("Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"\nTest accuracy: {test_acc:.4f}   Test loss: {test_loss:.4f}")

def analyze_curves(train_acc, val_acc, train_loss, val_loss):
    ta = train_acc[-1]
    va = val_acc[-1]
    tl = train_loss[-1]
    vl = val_loss[-1]
    msg = []
    if ta - va > 0.06:
        msg.append("The model shows signs of overfitting (training accuracy is noticeably higher than validation accuracy).")
    if ta < 0.70 and va < 0.70 and abs(ta-va) < 0.05:
        msg.append("The model may be underfitting (both train & val accuracies are low and similar).")
    if not msg:
        msg.append("No strong signs of severe overfitting or underfitting — training & validation curves are reasonably close.")
    suggestions = [
        "If overfitting: increase augmentation, add dropout, add weight decay, or reduce model capacity.",
        "If underfitting: increase model capacity, train longer, reduce regularization, or check for bugs in preprocessing.",
        "Also try learning-rate schedules, more data, or transfer learning from a pre-trained model."
    ]
    return "\n".join(msg) + "\n\nSuggestions:\n- " + "\n- ".join(suggestions)

analysis = analyze_curves(acc, val_acc, loss, val_loss)
print("\n=== Quick analysis of fit ===")
print(analysis)
```

```python
# FAST TRAINING PIPELINE (drop-in replacement)
# 1) Make sure your Colab runtime has GPU: Runtime -> Change runtime type -> GPU
from google.colab import drive
drive.mount('/content/drive')

import os, numpy as np, tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import itertools
from sklearn.utils import class_weight
tf.random.set_seed(42)

# --- Settings tuned for speed ---
DATASET_DIR = "/content/drive/MyDrive/COVID-19_Radiography_Dataset"  # update if needed
IMG_SIZE = 160           # smaller than 224; big speed win (try 128 or 160)
BATCH_SIZE = 32         # larger batch is faster if GPU memory allows
AUTOTUNE = tf.data.AUTOTUNE
VAL_SPLIT = 0.15
SEED = 42

# --- Mixed precision (fast on supported GPUs) ---
try:
    tf.keras.mixed_precision.set_global_policy('mixed_float16')
    print("Mixed precision enabled.")
except Exception as e:
    print("Mixed precision not enabled:", e)

# --- Build tf.data datasets (fast) ---
train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    DATASET_DIR,
    labels="inferred",
    label_mode="int",
    validation_split=VAL_SPLIT,
    subset="training",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)
val_ds = tf.keras.preprocessing.image_dataset_from_directory(
    DATASET_DIR,
    labels="inferred",
    label_mode="int",
    validation_split=VAL_SPLIT,
    subset="validation",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)
CLASS_NAMES = train_ds.class_names
NUM_CLASSES = len(CLASS_NAMES)
print("Classes:", CLASS_NAMES)

# compute class weights quickly for imbalance
all_labels = np.concatenate([y.numpy() for x,y in train_ds.unbatch().batch(1000000)])
cw = class_weight.compute_class_weight('balanced', classes=np.unique(all_labels), y=all_labels)
class_weights = {i: float(w) for i,w in enumerate(cw)}
print("Class weights:", class_weights)

# Fast augmentation inside tf.data (vectorized)
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.06),
    layers.RandomTranslation(0.04,0.04),
], name="fast_augmentation")

# Use preprocess_input corresponding to chosen backbone
from tensorflow.keras.applications import efficientnet
preprocess_input = efficientnet.preprocess_input  # for EfficientNetB0

def prepare(ds, augment=False):
    def _prep(image, label):
        image = tf.cast(image, tf.float32)
        image = preprocess_input(image)  # built-in normalization
        if augment:
            image = data_augmentation(image)
        return image, label
    return ds.map(_prep, num_parallel_calls=AUTOTUNE).cache().shuffle(1024).prefetch(AUTOTUNE)

train_ds_prepared = prepare(train_ds, augment=True)
val_ds_prepared = prepare(val_ds, augment=False)

# --- Choose backbone: EfficientNetB0 (fast + accurate) OR MobileNetV2 (very fast)
BACKBONE = "EfficientNetB0"  # change to "MobileNetV2" if you need more speed

if BACKBONE == "EfficientNetB0":
    base_model = tf.keras.applications.EfficientNetB0(
        include_top=False, weights="imagenet", input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )
else:
    base_model = tf.keras.applications.MobileNetV2(
        include_top=False, weights="imagenet", input_shape=(IMG_SIZE, IMG_SIZE, 3)
    )

base_model.trainable = False  # freeze for fastest initial training

# --- Build a small classifier head (keep dtype float32 for final outputs to avoid mixed precision softmax issues) ---
inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.2)(x)
# final layer: force float32 dtype so softmax works reliably with mixed precision
outputs = layers.Dense(NUM_CLASSES, activation="softmax", dtype='float32')(x)
model = keras.Model(inputs, outputs)

# --- Compile with an optimizer that works well with mixed precision ---
opt = tf.keras.optimizers.Adam(learning_rate=1e-3)
model.compile(optimizer=opt, loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()

# --- Callbacks: aggressive early stopping to save time ---
callbacks = [
    keras.callbacks.EarlyStopping(monitor="val_loss", patience=4, restore_best_weights=True),
    keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2)
]

# --- Train only the head (fast) ---
history = model.fit(
    train_ds_prepared,
    validation_data=val_ds_prepared,
    epochs=12,                # fewer epochs for speed; you can increase later
    class_weight=class_weights,
    callbacks=callbacks
)

# --- Optional: very small fine-tune for 1-2 epochs only (if you want small boost) ---
DO_FINETUNE = True
if DO_FINETUNE:
    base_model.trainable = True
    # unfreeze only last few layers for minimal compute
    for layer in base_model.layers[:-20]:
        layer.trainable = False
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-5), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    # train 1-3 epochs only (fast)
    history_ft = model.fit(train_ds_prepared, validation_data=val_ds_prepared, epochs=2, class_weight=class_weights, callbacks=callbacks)

# --- Quick evaluation on validation split ---
val_images = []
val_labels = []
for imgs, labels in val_ds.unbatch().batch(1000000):
    val_images = imgs.numpy(); val_labels = labels.numpy()
val_preds = model.predict(preprocess_input(val_images.astype("float32")), batch_size=64)
y_pred = np.argmax(val_preds, axis=1)
print(classification_report(val_labels, y_pred, target_names=CLASS_NAMES, digits=4))


```
